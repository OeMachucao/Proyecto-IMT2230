```python
# Importe de librerias necesarias
import json
import pandas as pd
from newspaper import Article
import time # Para no saturar los servidores al descargar
```

```python
with open('../data/fuentes.json', 'r') as f:
  biblioteca = json.load(f)

corpus_acumulado = []
# Ahora empezamos a hacer el loop de extracción de los links
# Recorremos cada pilar (Corporativo, Crítico, Prensa) y descargamos sus links
for pilar, urls in biblioteca.items():
    print(f"\n Procesando pilar: {pilar}")

    for url in urls:
      try:
        # Usamos newspaper3k para "limpiar" el HTML y quedarnos solo con texto
        articulo = Article(url)
        articulo.download()
        articulo.parse()

        # Guardamos un diccionario por cada documento
        # Esto será una fila en nuestro DataFrame
        corpus_acumulado.append({
                "pilar": pilar,
                "titulo": articulo.title,
                "texto": articulo.text,
                "url": url,
                "fuente": articulo.source_url
            })
        print(f" Éxito: {articulo.title[:50]}...")

        # REGLA DE ORO: No satures el servidor. Un segundo de descanso.
        time.sleep(1)
      except Exception as e:
        print(f" ERROR: Error en {url}: {e}")

# Converti en Dataframe y Exportar
df_final = pd.DataFrame(corpus_acumulado)
df_final.to_csv("../data/corpus_v1.csv", index=False)
print(f"\n ¡Misión cumplida! Tienes {len(df_final)} documentos listos para la SVD.")

```
